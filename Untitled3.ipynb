Aim
To demonstrate Locard's Exchange Principle by creating and modifying a file and observing the changes in its metadata (creation, modification, and access timestamps) using a Python program.bold text

Algorithm
Step 1: Start the program.

Step 2: Import the required modules (os, time, and datetime).

Step 3: Create a text file named evidence.txt.

Step 4: Write sample data into the file.

Step 5: Retrieve and display the file's creation time, modification time, and access time.

Step 6: Wait for 5 seconds using the sleep() function.

Step 7: Modify the file by appending additional text.

Step 8: Retrieve and display the updated file timestamps.

Step 9: Compare the timestamps before and after modification.

Step 10: Stop the program.



```python
import os
import time
from datetime import datetime

# File name
filename = "evidence.txt"

# Function to display file timestamps
def display_metadata():
    created = os.path.getctime(filename)
    modified = os.path.getmtime(filename)
    accessed = os.path.getatime(filename)

    print("\nFile Metadata")
    print("Created Time   :", datetime.fromtimestamp(created))
    print("Modified Time  :", datetime.fromtimestamp(modified))
    print("Accessed Time  :", datetime.fromtimestamp(accessed))

# Create the file
with open(filename, "w") as f:
    f.write("Original Evidence")

print("File Created Successfully")

# Show metadata before modification
print("\nBefore Modification")
display_metadata()

# Wait for 5 seconds
time.sleep(5)

# Modify the file
with open(filename, "a") as f:
    f.write("\nEvidence Modified")

print("\nFile Modified Successfully")

# Show metadata after modification
print("\nAfter Modification")
display_metadata()
```

Result
The Python program was executed successfully. The metadata of the file was displayed before and after modification
