Aim
To develop a Python program that generates a structured report describing the five stages of the Digital Forensic Investigation Lifecycle.

Algorithm
Start the program.

Create a list containing the five forensic investigation stages.

Display each stage with its description.

Save the report into a text file.

Display a success message.

Stop.

```python
stages = {
    "1. Identification": "Identify the incident and collect initial information.",
    "2. Preservation": "Preserve digital evidence without modification.",
    "3. Collection": "Collect digital evidence using forensic methods.",
    "4. Examination & Analysis": "Analyze the collected evidence.",
    "5. Reporting": "Prepare and present the forensic report."
}

with open("forensic_report.txt", "w") as file:
    file.write("DIGITAL FORENSIC INVESTIGATION REPORT\n\n")
    for stage, desc in stages.items():
        file.write(f"{stage}\n{desc}\n\n")
    print("Forensic report generated successfully.")
```
