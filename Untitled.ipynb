Verify Digital Evidence Using SHA-256 Hashing
Aim
To verify the integrity of digital evidence by generating and comparing SHA-256 hash values before and after acquisition.

Algorithm
Start the program.

Accept the file name from the user.

Open the file in binary mode.

Generate the SHA-256 hash value.

Display the generated hash.

Repeat the process after copying/imaging the file.

Compare both hash values.

If both hashes are identical, the evidence integrity is maintained.

Otherwise, report that the evidence has been modified.

Stop.



```python
import hashlib

def calculate_hash(filename):
    sha256 = hashlib.sha256()
    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)
    return sha256.hexdigest()

filename = input("Enter file name: ")

hash1 = calculate_hash(filename)

print("\nSHA-256 Hash:")
print(hash1)
```

Result
The SHA-256 hash values were compared successfully, and the program correctly determined whether the two files were identical or modified.
